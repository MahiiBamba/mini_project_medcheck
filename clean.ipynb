```python
import os
import shutil

# CHANGE THIS to your labels folder
folder = r"/home/mansi/Desktop/mini_project/incremntal_dataset/lables"

# Create a backup folder
backup_folder = folder + "_backup"

if not os.path.exists(backup_folder):
    shutil.copytree(folder, backup_folder)
    print("Backup created:", backup_folder)
else:
    print("Backup already exists:", backup_folder)


# Process files
for filename in os.listdir(folder):

    if not filename.lower().endswith(".txt"):
        continue

    name = os.path.splitext(filename)[0].lower()

    # k-* files: 0 -> 1, 1 -> 0
    if name.startswith("k-"):
        conversion = {
            "0": "1",
            "1": "0"
        }

    # v-* files: 15 -> 1, 16 -> 0
    elif name.startswith("v-"):
        conversion = {
            "15": "1",
            "16": "0"
        }

    else:
        continue

    filepath = os.path.join(folder, filename)

    with open(filepath, "r") as f:
        lines = f.readlines()

    new_lines = []

    for line in lines:
        parts = line.strip().split()

        if not parts:
            continue

        class_id = parts[0]

        if class_id in conversion:
            parts[0] = conversion[class_id]

        new_lines.append(" ".join(parts))

    with open(filepath, "w") as f:
        f.write("\n".join(new_lines) + "\n")

    print("Processed:", filename)

print("\n✅ Conversion completed!")
```

```python
import os
import shutil
import random

# ============================================================
# YOUR PATHS
# ============================================================

old_dataset = r"/home/mansi/Desktop/mini_project/blister_dataset_v2"
new_dataset = r"/home/mansi/Desktop/mini_project/incremntal_dataset"

output_dataset = r"/home/mansi/Desktop/mini_project/MedCheck_Combined"

# ============================================================
# CREATE OUTPUT FOLDERS
# ============================================================

folders = [
    os.path.join(output_dataset, "images", "train"),
    os.path.join(output_dataset, "images", "val"),
    os.path.join(output_dataset, "labels", "train"),
    os.path.join(output_dataset, "labels", "val")
]

for folder in folders:
    os.makedirs(folder, exist_ok=True)


# ============================================================
# COLLECT IMAGE + LABEL PAIRS
# ============================================================

pairs = []

def collect_pairs(image_folder, label_folder, dataset_name):

    for filename in os.listdir(image_folder):

        if not filename.lower().endswith(
            (".jpg", ".jpeg", ".png", ".bmp")
        ):
            continue

        image_path = os.path.join(image_folder, filename)

        # Remove image extension and add .txt
        base_name = os.path.splitext(filename)[0]
        label_filename = base_name + ".txt"

        label_path = os.path.join(label_folder, label_filename)

        if not os.path.exists(label_path):
            print("⚠️ Missing label:", filename)
            continue

        pairs.append(
            (image_path, label_path, filename, dataset_name)
        )


# ============================================================
# OLD DATASET
# ============================================================

collect_pairs(
    os.path.join(old_dataset, "images", "train"),
    os.path.join(old_dataset, "labels", "train"),
    "old_train"
)

collect_pairs(
    os.path.join(old_dataset, "images", "val"),
    os.path.join(old_dataset, "labels", "val"),
    "old_val"
)


# ============================================================
# NEW DATASET
# IMPORTANT: folder is called "lables"
# ============================================================

collect_pairs(
    os.path.join(new_dataset, "images"),
    os.path.join(new_dataset, "lables"),
    "new"
)


print("\n===================================")
print("TOTAL IMAGE-LABEL PAIRS:", len(pairs))
print("===================================")
```

```python
# Shuffle the 459 image-label pairs
random.seed(42)
random.shuffle(pairs)

# 80% train, 20% validation
train_count = int(len(pairs) * 0.8)

train_pairs = pairs[:train_count]
val_pairs = pairs[train_count:]

print("Total:", len(pairs))
print("Train:", len(train_pairs))
print("Validation:", len(val_pairs))
```

```python
def copy_pairs(pair_list, split):

    image_destination = os.path.join(
        output_dataset, "images", split
    )

    label_destination = os.path.join(
        output_dataset, "labels", split
    )

    for image_path, label_path, filename, dataset_name in pair_list:

        base_name = os.path.splitext(filename)[0]

        destination_image = os.path.join(
            image_destination, filename
        )

        destination_label = os.path.join(
            label_destination, base_name + ".txt"
        )

        # Handle duplicate image filenames
        if os.path.exists(destination_image):

            new_filename = dataset_name + "_" + filename
            new_label = dataset_name + "_" + base_name + ".txt"

            destination_image = os.path.join(
                image_destination, new_filename
            )

            destination_label = os.path.join(
                label_destination, new_label
            )

        shutil.copy2(image_path, destination_image)
        shutil.copy2(label_path, destination_label)


# Copy the two splits
copy_pairs(train_pairs, "train")
copy_pairs(val_pairs, "val")

print("✅ Combined dataset created!")
```

```python
import os
import cv2
from pathlib import Path

# ============================================================
# PATHS
# ============================================================

DATASET = Path("/home/mansi/Desktop/mini_project/MedCheck_Combined")

IMAGE_DIRS = {
    "train": DATASET / "images" / "train",
    "val": DATASET / "images" / "val"
}

LABEL_DIRS = {
    "train": DATASET / "labels" / "train",
    "val": DATASET / "labels" / "val"
}

# Output folder for visual inspection
OUTPUT_DIR = DATASET.parent / "MedCheck_Annotated_Check"

# ============================================================
# CLASS NAMES
# CHANGE THESE IF YOUR LABEL MEANINGS ARE DIFFERENT
# ============================================================

CLASS_NAMES = {
    0: "intact",
    1: "broken"
}

# ============================================================
# SUPPORTED IMAGE FORMATS
# ============================================================

IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

# ============================================================
# STATISTICS
# ============================================================

total_images = 0
total_annotations = 0

class_counts = {class_id: 0 for class_id in CLASS_NAMES}

missing_labels = []
empty_labels = []
extra_labels = []
invalid_annotations = []
images_without_boxes = []


# ============================================================
# DRAW YOLO BOX
# ============================================================

def draw_yolo_box(image, annotation, image_name):

    h, w = image.shape[:2]

    parts = annotation.strip().split()

    # YOLO format must contain exactly:
    # class_id x_center y_center width height
    if len(parts) != 5:
        return False, None

    try:
        class_id = int(parts[0])

        x_center = float(parts[1])
        y_center = float(parts[2])
        box_width = float(parts[3])
        box_height = float(parts[4])

    except ValueError:
        return False, None

    # Check class
    if class_id not in CLASS_NAMES:
        return False, None

    # Check YOLO coordinates
    values = [x_center, y_center, box_width, box_height]

    if not all(0 <= v <= 1 for v in values):
        return False, None

    if box_width <= 0 or box_height <= 0:
        return False, None

    # Convert YOLO coordinates to pixels
    x_center *= w
    y_center *= h
    box_width *= w
    box_height *= h

    x1 = int(x_center - box_width / 2)
    y1 = int(y_center - box_height / 2)

    x2 = int(x_center + box_width / 2)
    y2 = int(y_center + box_height / 2)

    # Clip to image boundaries
    x1 = max(0, min(x1, w - 1))
    y1 = max(0, min(y1, h - 1))
    x2 = max(0, min(x2, w - 1))
    y2 = max(0, min(y2, h - 1))

    if x1 >= x2 or y1 >= y2:
        return False, None

    # Draw rectangle
    cv2.rectangle(
        image,
        (x1, y1),
        (x2, y2),
        (0, 255, 0),
        2
    )

    # Label
    label = f"{class_id}: {CLASS_NAMES[class_id]}"

    # Background for text
    (text_w, text_h), baseline = cv2.getTextSize(
        label,
        cv2.FONT_HERSHEY_SIMPLEX,
        0.6,
        2
    )

    text_y = max(y1, text_h + 5)

    cv2.rectangle(
        image,
        (x1, text_y - text_h - baseline),
        (x1 + text_w, text_y),
        (0, 255, 0),
        -1
    )

    cv2.putText(
        image,
        label,
        (x1, text_y - 3),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.6,
        (0, 0, 0),
        2
    )

    return True, class_id


# ============================================================
# PROCESS TRAIN / VAL
# ============================================================

for split in ["train", "val"]:

    image_dir = IMAGE_DIRS[split]
    label_dir = LABEL_DIRS[split]

    output_dir = OUTPUT_DIR / split
    output_dir.mkdir(parents=True, exist_ok=True)

    print(f"\nProcessing {split.upper()}...")

    # --------------------------------------------------------
    # Find images
    # --------------------------------------------------------

    image_files = [
        p for p in image_dir.iterdir()
        if p.suffix.lower() in IMAGE_EXTENSIONS
    ]

    image_stems = {p.stem for p in image_files}

    # --------------------------------------------------------
    # Find labels
    # --------------------------------------------------------

    label_files = list(label_dir.glob("*.txt"))
    label_stems = {p.stem for p in label_files}

    # Extra labels
    for stem in sorted(label_stems - image_stems):
        extra_labels.append(
            f"{split}: {stem}.txt"
        )

    # --------------------------------------------------------
    # Process each image
    # --------------------------------------------------------

    for image_path in sorted(image_files):

        total_images += 1

        stem = image_path.stem
        label_path = label_dir / f"{stem}.txt"

        image = cv2.imread(str(image_path))

        if image is None:
            print(f"Could not read: {image_path}")
            continue

        # ----------------------------------------------------
        # Missing label
        # ----------------------------------------------------

        if not label_path.exists():

            missing_labels.append(
                f"{split}: {image_path.name}"
            )

            # Save original image to visualization folder
            cv2.imwrite(
                str(output_dir / image_path.name),
                image
            )

            continue

        # ----------------------------------------------------
        # Read annotations
        # ----------------------------------------------------

        with open(label_path, "r") as f:
            annotations = [
                line.strip()
                for line in f
                if line.strip()
            ]

        # Empty label file
        if len(annotations) == 0:

            empty_labels.append(
                f"{split}: {image_path.name}"
            )

            cv2.imwrite(
                str(output_dir / image_path.name),
                image
            )

            continue

        valid_boxes = 0

        # ----------------------------------------------------
        # Draw annotations
        # ----------------------------------------------------

        for annotation in annotations:

            valid, class_id = draw_yolo_box(
                image,
                annotation,
                image_path.name
            )

            if not valid:

                invalid_annotations.append(
                    f"{split}: {image_path.name} -> {annotation}"
                )

            else:

                valid_boxes += 1
                total_annotations += 1
                class_counts[class_id] += 1

        # ----------------------------------------------------
        # No valid boxes
        # ----------------------------------------------------

        if valid_boxes == 0:

            images_without_boxes.append(
                f"{split}: {image_path.name}"
            )

        # ----------------------------------------------------
        # Save visualized image
        # ----------------------------------------------------

        cv2.imwrite(
            str(output_dir / image_path.name),
            image
        )

    print(f"Finished {split.upper()}.")


# ============================================================
# REPORT
# ============================================================

report_path = OUTPUT_DIR / "annotation_report.txt"

with open(report_path, "w") as report:

    report.write("=" * 60 + "\n")
    report.write("MEDCHECK YOLO DATASET ANNOTATION REPORT\n")
    report.write("=" * 60 + "\n\n")

    report.write(f"Total images: {total_images}\n")
    report.write(f"Total annotations: {total_annotations}\n\n")

    report.write("CLASS DISTRIBUTION\n")
    report.write("-" * 30 + "\n")

    for class_id, count in class_counts.items():

        report.write(
            f"Class {class_id} ({CLASS_NAMES[class_id]}): "
            f"{count}\n"
        )

    report.write("\n")

    report.write("MISSING LABEL FILES\n")
    report.write("-" * 30 + "\n")

    if missing_labels:
        for item in missing_labels:
            report.write(item + "\n")
    else:
        report.write("None\n")

    report.write("\n")

    report.write("EMPTY LABEL FILES\n")
    report.write("-" * 30 + "\n")

    if empty_labels:
        for item in empty_labels:
            report.write(item + "\n")
    else:
        report.write("None\n")

    report.write("\n")

    report.write("EXTRA LABEL FILES\n")
    report.write("-" * 30 + "\n")

    if extra_labels:
        for item in extra_labels:
            report.write(item + "\n")
    else:
        report.write("None\n")

    report.write("\n")

    report.write("INVALID ANNOTATIONS\n")
    report.write("-" * 30 + "\n")

    if invalid_annotations:
        for item in invalid_annotations:
            report.write(item + "\n")
    else:
        report.write("None\n")

    report.write("\n")

    report.write("IMAGES WITHOUT VALID BOXES\n")
    report.write("-" * 30 + "\n")

    if images_without_boxes:
        for item in images_without_boxes:
            report.write(item + "\n")
    else:
        report.write("None\n")


# ============================================================
# FINAL OUTPUT
# ============================================================

print("\n" + "=" * 60)
print("DONE")
print("=" * 60)

print(f"\nVisualized dataset:")
print(OUTPUT_DIR)

print(f"\nReport:")
print(report_path)

print(f"\nTotal images: {total_images}")
print(f"Total annotations: {total_annotations}")

print("\nClass distribution:")

for class_id, count in class_counts.items():
    print(
        f"  {class_id} ({CLASS_NAMES[class_id]}): {count}"
    )

print(f"\nMissing labels: {len(missing_labels)}")
print(f"Empty labels: {len(empty_labels)}")
print(f"Extra labels: {len(extra_labels)}")
print(f"Invalid annotations: {len(invalid_annotations)}")
print(f"Images without valid boxes: {len(images_without_boxes)}")
```

```python
from pathlib import Path
import shutil

# ============================================================
# DATASET PATH
# ============================================================

DATASET = Path(
    "/home/mansi/Desktop/mini_project/MedCheck_Combined"
)

TRAIN_LABELS = DATASET / "labels" / "train"
VAL_LABELS = DATASET / "labels" / "val"


# ============================================================
# FILES THAT HAVE REVERSED CLASS LABELS
# m-7 through m-20
# ============================================================

files_to_fix = [
    f"m-{i}.txt"
    for i in range(7, 21)
]


# ============================================================
# FIND FILES IN TRAIN / VAL
# ============================================================

found_files = []

for filename in files_to_fix:

    train_file = TRAIN_LABELS / filename
    val_file = VAL_LABELS / filename

    train_exists = train_file.exists()
    val_exists = val_file.exists()

    # --------------------------------------------------------
    # File exists in BOTH folders -> STOP
    # --------------------------------------------------------

    if train_exists and val_exists:
        raise RuntimeError(
            f"\nERROR: {filename} exists in BOTH train and val.\n"
            "Script stopped without modifying anything."
        )

    # --------------------------------------------------------
    # Found in train
    # --------------------------------------------------------

    if train_exists:
        found_files.append(("train", train_file))

    # --------------------------------------------------------
    # Found in val
    # --------------------------------------------------------

    elif val_exists:
        found_files.append(("val", val_file))

    # --------------------------------------------------------
    # Not found
    # --------------------------------------------------------

    else:
        print(f"⚠️ NOT FOUND: {filename}")


# ============================================================
# SHOW WHAT WILL BE MODIFIED
# ============================================================

print("=" * 65)
print("MEDCHECK — CLASS LABEL CORRECTION")
print("=" * 65)

print("\nFiles that will be modified:\n")

for split, filepath in found_files:
    print(f"  {split:5} -> {filepath.name}")

print(f"\nTotal files found: {len(found_files)}")
print(f"Expected files:    {len(files_to_fix)}")


# ============================================================
# SAFETY CHECK
# ============================================================

if not found_files:
    raise RuntimeError(
        "\nNo target annotation files were found. "
        "Nothing was changed."
    )


print("\nThe following operation will be performed:")
print("  class 0  <->  class 1")
print("\nBounding-box coordinates will remain unchanged.")


# ============================================================
# CREATE BACKUPS
# ============================================================

print("\nCreating backups...\n")

for split, label_file in found_files:

    backup_dir = label_file.parent / "backup_before_class_swap"
    backup_dir.mkdir(exist_ok=True)

    backup_file = backup_dir / label_file.name

    # Don't overwrite an existing backup
    if backup_file.exists():
        raise RuntimeError(
            f"Backup already exists for {label_file.name}:\n"
            f"{backup_file}\n\n"
            "Script stopped to prevent overwriting the backup."
        )

    shutil.copy2(label_file, backup_file)

    print(f"  Backup: {split}/backup_before_class_swap/{label_file.name}")


# ============================================================
# SWAP CLASS IDS
# ============================================================

print("\nSwapping class IDs...\n")

for split, label_file in found_files:

    with open(label_file, "r") as f:
        lines = f.readlines()

    new_lines = []

    for line_number, line in enumerate(lines, start=1):

        stripped = line.strip()

        # Preserve empty lines
        if not stripped:
            new_lines.append(line)
            continue

        parts = stripped.split()

        # ----------------------------------------------------
        # Validate YOLO annotation
        # ----------------------------------------------------

        if len(parts) != 5:
            raise ValueError(
                f"\nInvalid YOLO annotation:\n"
                f"File: {label_file}\n"
                f"Line: {line_number}\n"
                f"Content: {stripped}\n\n"
                "Script stopped."
            )

        class_id = parts[0]

        # ----------------------------------------------------
        # Swap ONLY class ID
        # ----------------------------------------------------

        if class_id == "0":
            parts[0] = "1"

        elif class_id == "1":
            parts[0] = "0"

        else:
            raise ValueError(
                f"\nUnexpected class ID in:\n"
                f"File: {label_file}\n"
                f"Line: {line_number}\n"
                f"Class ID: {class_id}\n\n"
                "Expected only 0 or 1.\n"
                "Script stopped."
            )

        # Keep coordinates exactly as they were
        new_lines.append(" ".join(parts) + "\n")

    # --------------------------------------------------------
    # Write corrected file
    # --------------------------------------------------------

    with open(label_file, "w") as f:
        f.writelines(new_lines)

    print(f"  ✅ Fixed: {split}/{label_file.name}")


# ============================================================
# FINAL MESSAGE
# ============================================================

print("\n" + "=" * 65)
print("CORRECTION COMPLETE")
print("=" * 65)

print("\nModified files:")

for split, label_file in found_files:
    print(f"  {split:5} -> {label_file.name}")

print("\nClass IDs were swapped:")
print("  0 → 1")
print("  1 → 0")

print("\nBounding-box coordinates were NOT changed.")

print("\nOriginal files are backed up inside:")
print("  labels/train/backup_before_class_swap/")
print("  labels/val/backup_before_class_swap/")
```
